# Standard normal distribution

Area to the left of z = 1.80:

```r
pnorm(1.80)
```

Area to the right:

```r
1 - pnorm(1.80)
```

Middle probability:

```r
pnorm(1.5) - pnorm(-1.5)
```

Find the z-score at the 95th percentile:

```r
qnorm(0.95)
```

# Example: Race finishing times

Assume 10-km race times are normally distributed with:

- mean = 61 minutes
- standard deviation = 9 minutes

Probability a randomly selected runner takes more than 72 minutes:

```r
1 - pnorm(72, mean = 61, sd = 9)
```

Probability between 52 and 70 minutes:

```r
pnorm(70, mean = 61, sd = 9) -
  pnorm(52, mean = 61, sd = 9)
```

95th percentile:

```r
qnorm(0.95, mean = 61, sd = 9)
```
