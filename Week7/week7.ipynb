# Example: Race finishing times

Assume 10-km race times are normally distributed with:

- mean = 61 minutes
- standard deviation = 9 minutes

Probability a randomly selected runner takes more than 72 minutes:

```r
1 - pnorm(72, mean = 61, sd = 9)
```

Probability between 52 and 70 minutes:

```r
pnorm(70, mean = 61, sd = 9) -
  pnorm(52, mean = 61, sd = 9)
```

95th percentile:

```r
qnorm(0.95, mean = 61, sd = 9)
```

# In-Class Open Data Practice — Normal-Model Check with NYC Air Quality

**Source:** NYC Open Data, *Air Quality and Health Impacts*.

We will examine PM2.5 measurements for UHF34 neighborhoods. Instead of automatically assuming normality, we will compare the observed data with a normal-model calculation.

```r
url_air <- "https://data.cityofnewyork.us/resource/c3uy-2p5r.csv?$limit=50000"
air <- read.csv(url_air)

pm <- subset(
  air,
  name == "Fine particles (PM 2.5)" &
  geo_type_name == "UHF34"
)

pm$value <- as.numeric(pm$data_value)
pm <- pm[!is.na(pm$value), ]

mu_pm <- mean(pm$value)
sd_pm <- sd(pm$value)

mu_pm
sd_pm
hist(pm$value,
     main = "PM2.5 Measurements in NYC UHF34 Areas",
     xlab = "PM2.5 Data Value")
```

```r
# Use the empirical 75th percentile as a reference value
x <- as.numeric(quantile(pm$value, 0.75))

z <- (x - mu_pm) / sd_pm
normal_right <- 1 - pnorm(x, mean = mu_pm, sd = sd_pm)
empirical_right <- mean(pm$value > x)

x
z
normal_right
empirical_right
```
