# 1. Categorical data

A class survey asked students how they usually travel to campus.

```r
transport <- c(
  "Subway","Bus","Subway","Walk","Subway","Bus","Car","Subway",
  "Walk","Bus","Subway","Subway","Car","Bus","Walk","Subway",
  "Bus","Subway","Walk","Subway"
)

freq <- table(transport)
freq
```

Relative frequency:

```r
rel_freq <- prop.table(freq)
rel_freq
```

Bar chart:

```r
barplot(freq,
        main = "Transportation to Campus",
        xlab = "Transportation Method",
        ylab = "Frequency")
```

# 2. Quantitative data

# In-Class Open Data Practice — Organizing NYC 311 Data

**Source:** NYC Open Data, *311 Service Requests from 2020 to Present*.

Use frequency tables, relative frequencies, and graphs to summarize real service-request categories.

```r
url_311 <- "https://data.cityofnewyork.us/resource/erm2-nwe9.csv?$limit=5000"
nyc311 <- read.csv(url_311)

borough_data <- nyc311$borough
borough_data <- borough_data[borough_data != ""]

borough_freq <- table(borough_data)
borough_freq
prop.table(borough_freq)
```

```r
barplot(borough_freq,
        main = "NYC 311 Requests by Borough",
        xlab = "Borough",
        ylab = "Frequency")
```

```r
complaint_freq <- sort(table(nyc311$complaint_type), decreasing = TRUE)
top8 <- head(complaint_freq, 8)

top8
barplot(top8,
        las = 2,
        main = "Eight Most Common Complaint Types in This Sample",
        ylab = "Frequency")
```
