# 1. Categorical data

A class survey asked students how they usually travel to campus.

```r
transport <- c(
  "Subway","Bus","Subway","Walk","Subway","Bus","Car","Subway",
  "Walk","Bus","Subway","Subway","Car","Bus","Walk","Subway",
  "Bus","Subway","Walk","Subway"
)

freq <- table(transport)
freq
```

Relative frequency:

```r
rel_freq <- prop.table(freq)
rel_freq
```

Bar chart:

```r
barplot(freq,
        main = "Transportation to Campus",
        xlab = "Transportation Method",
        ylab = "Frequency")
```

# 2. Quantitative data

The following values are the numbers of minutes 30 students spent commuting.

```r
commute <- c(
  18,22,25,27,29,30,31,33,35,35,
  36,38,39,40,41,42,43,45,46,48,
  50,52,54,55,58,60,62,67,72,80
)
```

Create a histogram.

```r
hist(commute,
     main = "Student Commute Times",
     xlab = "Commute Time (minutes)",
     ylab = "Frequency")
```
