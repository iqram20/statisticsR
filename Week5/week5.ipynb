# In-Class Open Data Practice — Conditional Probability with NYC 311 Data

**Source:** NYC Open Data, *311 Service Requests from 2020 to Present*.

We will compare residential-noise complaints in Manhattan and Queens using a two-way table and conditional probabilities.

```r
url_311 <- "https://data.cityofnewyork.us/resource/erm2-nwe9.csv?$limit=10000"
nyc311 <- read.csv(url_311)

two_boroughs <- subset(
  nyc311,
  borough %in% c("MANHATTAN", "QUEENS")
)

two_boroughs$noise <- ifelse(
  two_boroughs$complaint_type == "Noise - Residential",
  "Residential Noise",
  "Other Complaint"
)

tab <- table(two_boroughs$borough, two_boroughs$noise)
addmargins(tab)
```

```r
# Conditional proportions within borough
prop.table(tab, margin = 1)

# Overall proportions
prop.table(tab)
```

```r
p_noise <- mean(two_boroughs$noise == "Residential Noise")
p_manhattan <- mean(two_boroughs$borough == "MANHATTAN")
p_noise_and_manhattan <- mean(
  two_boroughs$noise == "Residential Noise" &
  two_boroughs$borough == "MANHATTAN"
)

p_noise
p_manhattan
p_noise_and_manhattan
p_noise * p_manhattan
```
