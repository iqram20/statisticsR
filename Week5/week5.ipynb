# Example: Reading for fun

Suppose 500 children were surveyed.

```r
reading <- matrix(
  c(48,182,
    76,194),
  nrow = 2,
  byrow = FALSE,
  dimnames = list(
    Reading = c("Yes","No"),
    Gender = c("Boy","Girl")
  )
)

reading
```

Add totals.

```r
addmargins(reading)
```

Overall proportions:

```r
prop.table(reading)
```

Conditional distribution of reading status within each gender:

```r
prop.table(reading, margin = 2)
```
