# 1. A discrete probability distribution

A fair six-sided die is rolled once. If the result is odd, you win the number
of dollars shown. If the result is even, you lose that amount.

```r
x <- c(1,-2,3,-4,5,-6)
p <- rep(1/6, 6)

data.frame(Profit = x, Probability = p)
```

Expected value:

```r
sum(x * p)
```

# 2. Binomial probability

Suppose 20% of customers choose express delivery. If 12 customers are selected,
let X be the number who choose express delivery.

For a binomial model:

- fixed number of trials: n = 12,
- two outcomes per trial,
- constant probability p = 0.20,
- trials are independent.

Probability exactly 3 choose express delivery:

```r
dbinom(3, size = 12, prob = 0.20)
```

Probability at most 3:

```r
pbinom(3, size = 12, prob = 0.20)
```

Probability at least 3:

```r
1 - pbinom(2, size = 12, prob = 0.20)
```

Mean and standard deviation:

```r
n <- 12
p_success <- 0.20

mu <- n * p_success
sigma <- sqrt(n * p_success * (1 - p_success))

mu
sigma
```

# In-Class Open Data Practice — Building a Binomial Model from NYC 311 Data

**Source:** NYC Open Data, *311 Service Requests from 2020 to Present*.

Use the downloaded sample to estimate the probability that a request is a residential-noise complaint. Then use that estimated probability in a binomial model.

```r
url_311 <- "https://data.cityofnewyork.us/resource/erm2-nwe9.csv?$limit=10000"
nyc311 <- read.csv(url_311)

p_hat <- mean(nyc311$complaint_type == "Noise - Residential", na.rm = TRUE)
p_hat
```

```r
n <- 10

# Exactly 2 residential-noise complaints
dbinom(2, size = n, prob = p_hat)

# At most 2
pbinom(2, size = n, prob = p_hat)

# At least 2
1 - pbinom(1, size = n, prob = p_hat)

# Mean and standard deviation
n * p_hat
sqrt(n * p_hat * (1 - p_hat))
```
