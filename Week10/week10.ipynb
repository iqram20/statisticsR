# Example: Average wind speed

A researcher claims that the population mean wind speed is 8 mph.

A random sample of 32 days gives:

- x-bar = 8.2 mph
- s = 0.6 mph
- n = 32

We test:

H0: mu = 8

Ha: mu != 8

at alpha = 0.05.

# Test statistic

```r
xbar <- 8.2
mu0 <- 8
s <- 0.6
n <- 32

t_stat <- (xbar - mu0) / (s / sqrt(n))
t_stat
```

Degrees of freedom:

```r
df <- n - 1
df
```

Two-tailed p-value:

```r
p_value <- 2 * (1 - pt(abs(t_stat), df))
p_value
```

# Decision

```r
alpha <- 0.05

if (p_value < alpha) {
  print("Reject H0")
} else {
  print("Do not reject H0")
}
```

# In-Class Open Data Practice — Hypothesis Test with NYC Crash Data

**Source:** NYC Open Data, *Motor Vehicle Collisions - Crashes*.

For classroom practice, we will test a **hypothetical benchmark** of 0.30 persons injured per reported crash. The value 0.30 is a teaching benchmark, not an official NYC safety standard.

```r
url_crashes <- "https://data.cityofnewyork.us/resource/h9gi-nx95.csv?$limit=5000"
crashes <- read.csv(url_crashes)

injured <- as.numeric(crashes$number_of_persons_injured)
injured <- injured[!is.na(injured)]

set.seed(120)
sample_injured <- sample(injured, 100, replace = FALSE)

mean(sample_injured)
sd(sample_injured)
length(sample_injured)
```

We test:

**H0:** μ = 0.30

**Ha:** μ ≠ 0.30

Use α = 0.05.

```r
xbar <- mean(sample_injured)
s <- sd(sample_injured)
n <- length(sample_injured)
mu0 <- 0.30

t_stat <- (xbar - mu0) / (s / sqrt(n))
df <- n - 1
p_value <- 2 * (1 - pt(abs(t_stat), df))

t_stat
df
p_value
```

```r
t.test(
  sample_injured,
  mu = 0.30,
  alternative = "two.sided"
)
```
