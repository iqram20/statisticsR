# Example: Average wind speed

A researcher claims that the population mean wind speed is 8 mph.

A random sample of 32 days gives:

- x-bar = 8.2 mph
- s = 0.6 mph
- n = 32

We test:

H0: mu = 8

Ha: mu != 8

at alpha = 0.05.

# Test statistic

```r
xbar <- 8.2
mu0 <- 8
s <- 0.6
n <- 32

t_stat <- (xbar - mu0) / (s / sqrt(n))
t_stat
```

Degrees of freedom:

```r
df <- n - 1
df
```

Two-tailed p-value:

```r
p_value <- 2 * (1 - pt(abs(t_stat), df))
p_value
```

# Decision

```r
alpha <- 0.05

if (p_value < alpha) {
  print("Reject H0")
} else {
  print("Do not reject H0")
}
```

# Using raw data with t.test()

```r
visits <- c(3,2,1,3,7,2,9,4,6,6,
            8,0,5,6,4,2,1,3,4,1)

mean(visits)
sd(visits)

t.test(visits,
       mu = 5.8,
       alternative = "less")
```
