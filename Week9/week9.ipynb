# Example: Newborn weights

Suppose the weights (lb) of 10 randomly selected newborns are:

```r
weights <- c(6.1, 6.8, 7.0, 7.2, 7.4,
             7.5, 7.7, 7.8, 8.1, 7.0)
```

Sample statistics:

```r
mean(weights)
sd(weights)
length(weights)
```

90% confidence interval:

```r
t.test(weights, conf.level = 0.90)$conf.int
```

# Manual calculation

```r
xbar <- mean(weights)
s <- sd(weights)
n <- length(weights)
df <- n - 1

t_star <- qt(1 - 0.10/2, df)
margin_error <- t_star * s / sqrt(n)

lower <- xbar - margin_error
upper <- xbar + margin_error

t_star
margin_error
c(lower, upper)
```

# Sample-size planning when sigma is known

Suppose we want to estimate a population mean within E = 20 units,
using sigma = 63 and 95% confidence.

```r
z_star <- qnorm(0.975)
sigma <- 63
E <- 20

n_required <- (z_star * sigma / E)^2
n_required
ceiling(n_required)
```

# In-Class Open Data Practice — Confidence Interval with NYC Crash Data

**Source:** NYC Open Data, *Motor Vehicle Collisions - Crashes*.

We will take a random sample of reported crashes and estimate the mean number of persons injured per crash.

```r
url_crashes <- "https://data.cityofnewyork.us/resource/h9gi-nx95.csv?$limit=5000"
crashes <- read.csv(url_crashes)

injured <- as.numeric(crashes$number_of_persons_injured)
injured <- injured[!is.na(injured)]

set.seed(120)
sample_injured <- sample(injured, 60, replace = FALSE)

mean(sample_injured)
sd(sample_injured)
length(sample_injured)
```

```r
# 90% t confidence interval
t.test(sample_injured, conf.level = 0.90)$conf.int
```

```r
# Manual calculation
xbar <- mean(sample_injured)
s <- sd(sample_injured)
n <- length(sample_injured)
df <- n - 1

t_star <- qt(1 - 0.10/2, df)
ME <- t_star * s / sqrt(n)

lower <- xbar - ME
upper <- xbar + ME

t_star
ME
c(lower, upper)
```
