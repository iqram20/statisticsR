# Example: Test scores

```r
scores <- c(
  31,33,35,36,40,40,41,43,45,47,
  48,51,51,52,55,56,59,70,71,74,
  76,78,84,87,87,90,93
)
```

Summary statistics:

```r
mean(scores)
median(scores)
sd(scores)
fivenum(scores)
summary(scores)
```

Histogram:

```r
hist(scores,
     main = "Test Score Distribution",
     xlab = "Score")
```

Boxplot:

```r
boxplot(scores,
        horizontal = TRUE,
        main = "Boxplot of Test Scores",
        xlab = "Score")
```

# In-Class Open Data Practice — NYC Motor Vehicle Collisions

**Source:** NYC Open Data, *Motor Vehicle Collisions - Crashes*.

We will summarize the number of persons injured in reported crashes using measures of center, spread, quartiles, a histogram, and a boxplot.

```r
url_crashes <- "https://data.cityofnewyork.us/resource/h9gi-nx95.csv?$limit=5000"
crashes <- read.csv(url_crashes)

injured <- as.numeric(crashes$number_of_persons_injured)
injured <- injured[!is.na(injured)]

length(injured)
head(injured)
```

```r
mean(injured)
median(injured)
sd(injured)
quantile(injured)
fivenum(injured)
```

```r
hist(injured,
     main = "Number of Persons Injured per Reported Crash",
     xlab = "Persons Injured")

boxplot(injured,
        horizontal = TRUE,
        main = "Persons Injured per Reported Crash",
        xlab = "Persons Injured")
```
