# Example: Test scores

```r
scores <- c(
  31,33,35,36,40,40,41,43,45,47,
  48,51,51,52,55,56,59,70,71,74,
  76,78,84,87,87,90,93
)
```

Summary statistics:

```r
mean(scores)
median(scores)
sd(scores)
fivenum(scores)
summary(scores)
```

Histogram:

```r
hist(scores,
     main = "Test Score Distribution",
     xlab = "Score")
```

Boxplot:

```r
boxplot(scores,
        horizontal = TRUE,
        main = "Boxplot of Test Scores",
        xlab = "Score")
```
