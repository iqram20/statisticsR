# Create a population

We create a positively skewed population.

```r
set.seed(120)
population <- rexp(10000, rate = 1/20)

hist(population,
     main = "Population Distribution",
     xlab = "Value")
```

Population mean and standard deviation:

```r
mean(population)
sd(population)
```

# One random sample

```r
samp <- sample(population, 25)
mean(samp)
hist(samp)
```

# Sampling distribution of the mean

Generate 5000 random samples of size 25 and store each sample mean.

```r
sample_means_25 <- replicate(
  5000,
  mean(sample(population, 25))
)

hist(sample_means_25,
     main = "Sampling Distribution: n = 25",
     xlab = "Sample Mean")
```

Now repeat with n = 100.

```r
sample_means_100 <- replicate(
  5000,
  mean(sample(population, 100))
)

hist(sample_means_100,
     main = "Sampling Distribution: n = 100",
     xlab = "Sample Mean")
```

Compare standard deviations:

```r
sd(sample_means_25)
sd(sample_means_100)
```

```r
mu <- 16
sigma <- 0.6
n <- 25

se <- sigma / sqrt(n)
se

pnorm(15.8, mean = mu, sd = se)
```

# In-Class Open Data Practice — Sampling Distributions with NYC Air Quality

**Source:** NYC Open Data, *Air Quality and Health Impacts*.

Treat the available PM2.5 UHF34 measurements as a finite classroom population. We will repeatedly draw random samples and study the distribution of their means.

```r
url_air <- "https://data.cityofnewyork.us/resource/c3uy-2p5r.csv?$limit=50000"
air <- read.csv(url_air)

pm <- subset(
  air,
  name == "Fine particles (PM 2.5)" &
  geo_type_name == "UHF34"
)

population <- as.numeric(pm$data_value)
population <- population[!is.na(population)]

mean(population)
sd(population)
length(population)
```

```r
set.seed(120)

means_10 <- replicate(
  3000,
  mean(sample(population, 10, replace = FALSE))
)

means_30 <- replicate(
  3000,
  mean(sample(population, 30, replace = FALSE))
)

hist(means_10,
     main = "Sampling Distribution of Mean: n = 10",
     xlab = "Sample Mean")

hist(means_30,
     main = "Sampling Distribution of Mean: n = 30",
     xlab = "Sample Mean")

sd(means_10)
sd(means_30)
```

```r
pop_sd <- sd(population)

theoretical_se_10 <- pop_sd / sqrt(10)
theoretical_se_30 <- pop_sd / sqrt(30)

theoretical_se_10
theoretical_se_30
```
